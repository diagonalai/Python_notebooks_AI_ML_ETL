```python
from getpass import getpass
from google import genai

#CREATE FREE API KEY here https://aistudio.google.com/app/api-keys
#Models; https://ai.google.dev/gemini-api/docs/models

# 1️⃣ Ask for API key at runtime
gemini_key = getpass("Enter your Gemini API Key: ")

# 2️⃣ Create client with your API key
client = genai.Client(api_key=gemini_key)

# 3️⃣ Send prompt
prompt = "Write a friendly greeting in Spanish."
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

# 4️⃣ Print result
print(response.text)
```
